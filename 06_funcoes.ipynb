## Solving a quadratic equation

In this example, we create a solver for the quadratic equation

\begin{equation}
\nonumber ax^2 + bx + c = 0
\end{equation}

The condition $a \neq 0$ is required for the equation to be quadratic. Its solutions are

\begin{equation}
\nonumber x = \frac{-b \pm \sqrt{b^2 - 4ac}}{2a}.
\end{equation}

For the roots to be real, the expression under the square root must be nonnegative. This quantity is the discriminant and is conventionally denoted by the uppercase Greek letter delta, $\Delta$:

\begin{equation}
\nonumber \Delta = b^2 - 4ac.
\end{equation}

```python
def eq_solver(a,b,c):
    discriminante = b**2 - 4*a*c
    if discriminante < 0: 
        return 'no real roots'
    else:
        print('Discriminant = ', discriminante)
        x1 = (-b + np.sqrt(discriminante))/(2*a)
        x2 = (-b - np.sqrt(discriminante))/(2*a)

        print('x1 =', x1)
        print('x2 =', x2)

        return x1, x2
```

```python
eq_solver(a = 1,b = 2,c = 1)
```

## More examples

```python
def comparador(x , y):
    if (x > y):
        print(x, 'greater than', y)
    elif (x < y):
        print(x, 'less than', y)
    else:
        print(x, 'equal to', y)
        
comparador(1,1)
comparador(10,1)
comparador(1,10)
```

```python
def compara_string(texto_1, texto_2):
    if (texto_1 == texto_2):
        print('The strings are equal! (', texto_1, ')')
    else:
        print('The strings are different!', '(', texto_1, ')', ' differs from ', '(', texto_2, ')')

compara_string(texto_1 = 'aaa', texto_2 = 'aaa')
compara_string(texto_1 = 'aaa', texto_2 = 'bb')
```
