## Lists

A list is an ordered sequence of items. It is one of the most commonly used and flexible data types in Python. The elements of a list do not need to have the same type. A list is declared by placing comma-separated items inside square brackets `[ ]`.

```python
# Define a list
lista = [1, 1, 2, 3, 5, 8, 13, 21]

# Access a specific element
print('This is the original list element [2]:', lista[2])

# List elements are mutable
lista[2] = 4
print('This is the new list element [2]:', lista[2])

# List elements may have different types
lista[2] = 'Python'
print('This is the new list element [2], now a string:', lista[2])

lista
```

## Tuples

A tuple is written using comma-separated items, usually enclosed in parentheses `( )`.

## Strings

A string is a sequence of Unicode characters. Single or double quotation marks can be used to represent strings, while multiline strings can be written with triple quotes.

```python
s = 'Hello world!'

print("s[4] = ", s[4])
print("s[6:11] = ", s[6:11])
```

```python
# Strings are also immutable in Python; assigning a new value to an indexed character raises an error.
s[5] ='d'
```

## Sets

A set is an unordered collection of unique items. A set literal is defined by comma-separated values inside braces `{ }`. Set elements do not have a positional order.

```python
a = {5,2,3,1,4}

print("a = ", a)
print(type(a))
```

```python
# Python sets are unordered and do not support positional indexing, so this expression raises an error.
a[1]
```

## Python Dictionaries

A dictionary is a collection of key–value pairs. It is commonly used when data should be retrieved by an identifier rather than by position. Dictionaries are optimized for key-based lookup. In Python, dictionary literals use braces `{ }`, with each item written as `key: value`. Keys and values may have different types, provided that keys are hashable.

```python
d = {1:'value','key':2}
print(type(d))

print("d[1] = ", d[1])
print("d['key'] = ", d['key'])
```

```python
# The code below raises an error because 2 is a value associated with the key 'key', not a key itself.
print("d[2] = ", d[2])
```

```python
# Existing dictionary values can be replaced.
d['key'] =  5

d['key']
```

## Converting between collection types

```python
# List example
lista = ['avocado','banana','persimmon','apricot', 'embauba','fig'] 
print('list:',lista)

# Tuple example
tupla = (1, 2, 3, 4, 5, 6, 7)
print('tuple:',tupla)

# Dictionary example
dicionario = {'a':'avocado', 'b':'ball', 'c':'dog', 'd':'die'} 
print('dictionary:',dicionario)

# Set example
conjunto = {'a', 'e', 'i' , 'o', 'u'} 
print('set:',conjunto)

# String example
string = 'Hello World!'
print('string:',string)
```

```python
# Convert a list to a set
lista_em_conjunto = set(lista)

print(type(lista))
print(type(lista_em_conjunto))

lista_em_conjunto
```

```python
# Convert a set to a tuple
conjunto_em_tupla = tuple(conjunto)

print(type(conjunto))
print(type(conjunto_em_tupla))

conjunto_em_tupla
```

```python
# Convert a string to a list
string_em_lista = list(string)

string_em_lista
```

```python
lista_1 = [1,'pineapple']
lista_2 = [2,'banana']
lista_3 = [3,'persimmon']

# Converting to a dictionary requires an iterable of key–value pairs
dicionario_forma_comum = {1:'pineapple', 2:'banana', 3:'persimmon'}

print('dicionario_forma_comum =',dicionario_forma_comum)
dicionario_forma_comum
```
