# Nested Functions

## Simple examples — sum of the first n natural numbers

```python
total = 0 
k = 1

n = int(input("Enter a positive integer:"))

while k <= n:
    total = total + k
    k = k + 1
    
print(total)
```

```python
def soma(n):
    total = 0 
    k = 1
    while k <= n:
        total = total + k
        k = k + 1
    return total

soma(100)
```

```python
def soma(n, term):
    total, k = 0, 1
    while k <= n:
        total, k = total + term(k), k + 1
    return total

def identidade(x):
    return x

soma(100, identidade)
```

```python
def soma(n, term):
    total, k = 0, 1
    while k <= n:
        total, k = total + term(k), k + 1
    return total

def identidade(x):
    return x

def soma_n_final(n):
    return soma(n, identidade)

soma_n_final(100)
```

## Nested functions with parameters

Arguments can be passed to inner functions.

```python
def saudacao(nome):
    def mensagem():
        return f"Hello, {nome}! Welcome to the Python course."
    
    return mensagem()

print(saudacao("Maria"))
```

## Nested function with a closure
An inner function forms a closure when it retains access to variables from the enclosing function even after the enclosing function has finished executing.

```python
def multiplicador(n):
    def multiplica(x):
        return x * n  # Retains the value of n after the outer function finishes executing
    return multiplica

dobro = multiplicador(2)
triplo = multiplicador(3)

x = 5
print(f'Double of {x} =', dobro(x))  # 10
print(f'Triple of {x} =', triplo(x))  # 15
```
