# Conditionals (`if`, `elif`, and `else`)

## The `if` statement

An `if` statement executes an indented block only when its condition evaluates to `True`. A colon follows the condition, and indentation defines the block.

```python
num = float(input('Enter a number:'))

if num > 0:
    print(num, 'is a positive number.')

num
```

```python
a = int(input("Enter an integer: "))
if (a % 2) == 1:
    print("Odd number")
```

```python
a = int(input("Enter an integer: "))
if (a % 2) == 0:
    print("Even number")
if (a % 2) == 1:
    print("Odd number")
```

## The `else` statement

An `else` block runs when the preceding `if` condition is false.

```python
a = int(input("Enter an integer: "))
if (a % 2) == 0:
    print("Even number")
else:
    print("Odd number")
```

```python
a = float(input("Enter the first number: "))
b = float(input("Enter the second number: "))
if a > b:
    print("The larger number is", a)
else:
    print("The larger number is", b)
```

```python
a = float(input("Enter the first number: "))
b = float(input("Enter the second number: "))
if a == b:
    print("The two numbers are equal")
else:
    if a > b:
        print("The first number is larger")
    else:
        print("The second number is larger")
```

## The `elif` statement

`elif` introduces an additional condition that is tested when previous conditions were false.

```python
a = float(input("Enter the first number: "))
b = float(input("Enter the second number: "))
if a == b:
    print("The two numbers are equal")
elif a > b:
    print("The first number is larger")
else:
    print("The second number is larger")
```

```python
a = float(input('Enter a value for a:'))
b = float(input('Enter a value for b:'))
c = float(input('Enter a value for c:'))
delta = b**2 - 4*a*c
if delta == 0:
    print('delta is equal to zero')
elif delta > 0:
    print('delta is greater than zero')
elif delta < 0:
    print('delta is less than zero')
print('The value of delta is', delta)
```

```python
a = float(input('Enter a value for a:'))
b = float(input('Enter a value for b:'))
c = float(input('Enter a value for c:'))
delta = b**2 - 4*a*c
if delta >= 0:
    print('The quadratic equation has real roots.')
elif delta < 0:
    print('The quadratic equation has no real roots.')
print('The value of delta is', delta)
```

## Compound conditionals

```python
a = float(input('Enter a value for a:'))
b = float(input('Enter a value for b:'))
c = float(input('Enter a value for c:'))
if a > b and b > c:
    print('a > c')
elif a < b and b < c:
    print('a < c')
elif a < b and b > c and a > c:
    print('a < b and a > c')
```
