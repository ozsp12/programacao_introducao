## Identity operators (`is` and `is not`)

```python
x1 = 5
y1 = 5
print(x1 is not y1)
```

```python
x2 = 'Hello'
y2 = 'Hello'
print(x2 is y2)
```

```python
# This comparison returns False because x3 and y3 are distinct list objects, unlike the immutable string and numeric examples above.
# Lists are discussed in more detail later.
x3 = [1,2,3]
y3 = [1,2,3]
print(x3 is y3)
```

```python
x = 'Hello World!'

print('W' in x)
print('w' in x)
print('!' in x)
print('' in x)
```

```python
print('hello' not in x)
print('Hello' in x)
```

```python
# Here we create a data structure that will be discussed later: a dictionary.
# This example maps numerical positions to letters of the alphabet.
y = { 1:'a', 2:'b', 3:'c', 4:'d', 5:'e', 6:'f', 7:'g', 8:'h', 9:'i', 10:'j',
     11:'l', 12:'m', 13:'n', 14:'o', 15:'p', 16:'q', 17:'r', 18:'s', 19:'t',
     20:'u', 21:'v', 22:'x', 23:'z'
    }

# Print the complete dictionary.
print(y)
```

```python
# Using a numerical key inside brackets retrieves the associated value from the dictionary.
print(y[1], y[12] ,y[23])
```

```python
print('w' in y)

# 'a' is a value rather than a key, so this expression returns False.
print('a' in y)

# The key 1 is present in the dictionary, so this expression returns True.
print(1 in y)
```
