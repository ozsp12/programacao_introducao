# Introduction to Dictionaries

Dictionaries are data structures composed of key–value pairs and implemented using hash tables. Each key is transformed by a hash function and used to locate its associated value. Key lookup, insertion, and update have average-case time complexity $O(1)$. In Python, keys must be immutable and hashable. Dictionaries are suitable for representing records, building indexes, mapping identifiers, and organizing data that does not depend on positional order. Their operation follows the general principles of hash tables described in classical data-structure algorithms. Python uses an optimized open-addressing hash-table design with probing and a compact internal storage array. The complete specification is available in the official Python documentation.

# Examples

## Working with Database Records

Suppose we want to build a dictionary in which each key is a unique identifier and each value is the record with the largest timestamp associated with that identifier. The expected result is:

```python
# Example data
data = [
 {"id": 1234, "timestamp": 1, "metadata": "a"},
 {"id": 1234, "timestamp": 2, "metadata": "b"},
 {"id": 2345, "timestamp": 4, "metadata": "c"},
 {"id": 2345, "timestamp": 2, "metadata": "d"},
 {"id": 5678, "timestamp": 1, "metadata": "e"}
]

# Expected result
expected_output = {
1234: {"id": 1234, "timestamp": 2, "metadata": "b"}, 2345: {"id": 2345, "timestamp": 4, "metadata": "c"},
 5678: {"id": 5678, "timestamp": 1, "metadata": "e"}
}

expected_output
```

```python
# Initialize the final dictionary
expected_output_answer = {}
# Auxiliary dictionary storing, for each id, the record with the largest timestamp found so far
auxiliary_dict = {}
# Iterate over each record in the list
for row in data:
    # Extract the current record id
    i = row['id']
    # Check whether the id is new or the current timestamp is larger than the stored timestamp
    if i not in auxiliary_dict or row['timestamp'] > auxiliary_dict[i]['timestamp']:
        # Update the record associated with this id
        auxiliary_dict[i] = row
# Assign the final result to the requested variable
expected_output_answer = auxiliary_dict
# Display the result
expected_output_answer
```

```python
expected_output_answer == expected_output
```

## Handling Missing Data

Consider a list of records containing `id`, `timestamp`, and `metadata`. Some records have `timestamp = None`, while every valid timestamp is known to be strictly positive. We want to build a dictionary whose key is the id and whose value is the record with the largest timestamp, treating `None` as smaller than any positive timestamp. In this example, `None` is replaced by 1 for comparison purposes.

```python
# List of records
data = [
 {"id": 1234, "timestamp": None, "metadata": "b"},
 {"id": 2345, "timestamp": 4,   "metadata": "c"},
 {"id": 2345, "timestamp": 2,   "metadata": "d"},
 {"id": 5678, "timestamp": 1,   "metadata": "e"}
]
# Display the input data
data
```

```python
# Auxiliary dictionary storing the best record for each id
auxiliary_dict = {}

# Normalize None values by assigning them the minimum comparison value
def correct_none(t):
    return float(1) if t is None else t

# Iterate over each record
for row in data:

    # Extract id and timestamp
    t = row['timestamp']
    i = row['id']

    # Compare normalized timestamps and retain the larger one
    if i not in auxiliary_dict or correct_none(t) > correct_none(auxiliary_dict[i]['timestamp']):
        auxiliary_dict[i] = row
        
# Final result
expected_output_answer = auxiliary_dict
expected_output_answer
```

## Applying a Business Rule to Metadata

Consider a list of records containing `id`, `timestamp`, and `metadata`. For each id, we want to select the maximum record according to the following comparison rule. Primary criterion: largest timestamp. Tie-breaking criterion: lexicographically largest metadata value when timestamps are equal. The objective is to build a dictionary in which each key is an id and each value is the record selected according to these rules.

```python
# Records to be processed
data = [
 {"id": 1234, "timestamp": 1, "metadata": "a"},
 {"id": 1234, "timestamp": 1, "metadata": "c"},
 {"id": 2345, "timestamp": 4, "metadata": "c"},
 {"id": 2345, "timestamp": 2, "metadata": "d"},
 {"id": 5678, "timestamp": 1, "metadata": "e"}
]
# Display the input data
data
```

```python
# Final dictionary
expected_output_answer = {}

# Auxiliary dictionary retaining the best record for each id
auxiliary_dict = {}

# Iterate over each record
for row in data:

    # Extract the current id
    i = row['id']

    # Check whether the id has not yet been stored,
    # whether the current timestamp is larger,
    # or, in a timestamp tie, whether the current metadata is larger
    if (i not in auxiliary_dict or
        row['timestamp'] > auxiliary_dict[i]['timestamp'] or
        row['metadata'] > auxiliary_dict[i]['metadata']):
        
        # Update the record associated with the id
        auxiliary_dict[i] = row

# Store the final result
expected_output_answer = auxiliary_dict

# Display the resulting dictionary
expected_output_answer
```
