# Increasing Sequence Check
Develop a program that reads a number $n$ followed by a sequence of $n$ numbers. The program must determine whether the sequence is sorted in increasing order.

```python
def verifica_crescente():
    n = int(input("Enter the number of elements: "))
    lista = [int(input()) for _ in range(n)]
    
    if lista == sorted(lista):
        print("The sequence is sorted in increasing order.")
    else:
        print("The sequence is NOT sorted.")

verifica_crescente()
```

# Greatest Common Divisor (GCD)
Write a program that reads two numbers $m$ and $n$ and computes their GCD using Euclid's algorithm.

```python
def mdc(m, n):
    while n:
        m, n = n, m % n
    return m

m = int(input("Enter m: "))
n = int(input("Enter n: "))
print("GCD:", mdc(m, n))
```

# Nearest Prime Numbers

Write a program that reads a number $n$ and prints the largest prime number less than or equal to $n$ and the smallest prime number greater than or equal to $n$.

```python
def eh_primo(num):
    if num < 2:
        return False
    for i in range(2, int(num**0.5) + 1):
        if num % i == 0:
            return False
    return True

def primo_proximo(n):
    menor = n
    while not eh_primo(menor):
        menor -= 1
    
    maior = n
    while not eh_primo(maior):
        maior += 1
    
    print(f"Largest prime ≤ {n}: {menor}")
    print(f"Smallest prime ≥ {n}: {maior}")

n = int(input("Enter n: "))
primo_proximo(n)
```

# Counting Values in Intervals

Develop a program that reads $n$ integers and counts how many fall within the intervals [0, 25], [26, 50], [51, 75], and [76, 100].

```python
def contagem_intervalos():
    n = int(input("Enter the number of values: "))
    valores = [int(input()) for _ in range(n)]
    
    contadores = [0] * 4
    for v in valores:
        if 0 <= v <= 25:
            contadores[0] += 1
        elif 26 <= v <= 50:
            contadores[1] += 1
        elif 51 <= v <= 75:
            contadores[2] += 1
        elif 76 <= v <= 100:
            contadores[3] += 1

    print("Intervals:")
    print(f"[0,25]: {contadores[0]}")
    print(f"[26,50]: {contadores[1]}")
    print(f"[51,75]: {contadores[2]}")
    print(f"[76,100]: {contadores[3]}")

contagem_intervalos()
```

# Square Root Calculation
Implement a program that computes the square root of a positive number using Newton's method of successive approximations and prints the twentieth approximation.

```python
def raiz_quadrada(n):
    x = n
    for _ in range(20):  # 20 Newton iterations
        x = 0.5 * (x + n / x)
    return x

n = float(input("Enter a number: "))
print(f"The approximate square root is: {raiz_quadrada(n)}")
```

# Decimal-to-Binary Conversion
Develop a program that reads a decimal integer and prints its binary representation.

# Numeric Pattern I
Create a program that reads a positive integer $n$ and prints $n$ lines forming a numerical right-triangle pattern.

```python
def padrao1(n):
    for i in range(1, n + 1):
        print(" ".join(str(x) for x in range(1, i + 1)))

n = int(input("Enter n: "))
padrao1(n)
```

# Numeric Pattern II
Write a program that reads a number $n$ and prints $n$ lines forming an inverted numerical right-triangle pattern.

```python
def padrao2(n):
    for i in range(n, 0, -1):
        print(" ".join(str(x) for x in range(i, 0, -1)))

n = int(input("Enter n: "))
padrao2(n)
```

# Symbol Pattern
Develop a program that reads a number $n$ and prints an $n \times n$ square whose diagonal contains `+` and whose remaining positions contain `*`.

```python
def padrao_simbolos(n):
    for i in range(n):
        linha = ["*"] * n  # Start with every position containing '*'
        linha[i] = "+"  # Set the corresponding diagonal position to '+'
        print(" ".join(linha))  # Join and print the formatted row

n = int(input("Enter the value of n: "))
padrao_simbolos(n)
```

# Mega-Sena Lottery Game Generation
Implement a program that generates possible Mega-Sena lottery selections according to a specified alternation rule between even and odd numbers.

```python
import itertools

def jogos_supersticiosos(n):
    # Numbers available in Mega-Sena (1 through 60)
    numeros = range(1, 61)

    # Separate even and odd numbers
    pares = [num for num in numeros if num % 2 == 0]
    impares = [num for num in numeros if num % 2 == 1]

    # The pattern is (E, O, E, O, E, O), so itertools.product can be used
    jogos_iter = itertools.product(pares, impares, pares, impares, pares, impares)

    # Display the first n generated selections
    print("Displaying the first selections:")
    for jogo in itertools.islice(jogos_iter, n):
        print(jogo)

    # Compute the total number of possible selections without generating all of them
    total_jogos = (len(pares) * len(impares)) ** 3  # three even and three odd positions
    print(f"\nTotal number of possible selections: {total_jogos}")

n = 20
jogos_supersticiosos(n)
```
