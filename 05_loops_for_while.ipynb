## `while`

```python
dividendo = int(input("Enter the dividend: "))
divisor = int(input("Enter the divisor: "))

quociente = 0

while dividendo >= divisor:
    dividendo = dividendo - divisor
    quociente = quociente + 1

print("Quotient:", quociente)
print("Remainder:", dividendo)
```

```python
base = float(input("Base: "))
exp = int(input("Exponent: "))

result = base
i = 1

while i < exp:
    result = result * base
    i = i + 1

print(result)
```

```python
base = float(input("Base: "))
exp = int(input("Exponent: "))

result = 1

while exp > 0:
    result = result * base
    exp = exp - 1

while exp < 0:
    result = result / base
    exp = exp + 1

print(result)
```

## `while` and lists

```python
letras = ['a','b','c','d','e','f','g','h']
i = 0

while i < len(letras):
    print(letras[i])
    i = i + 1
```

```python
# Find the maximum value in an arbitrary set of numbers
numeros = [11, 15, 21, 9, 14, 12,13, 100, 1000]
maximo = 0

i = 0
while i < len(numeros):
    if numeros[i] > maximo:
        maximo = numeros[i]
    i = i + 1
    
print(maximo)
```

## `for`

```python
# Find the maximum value in a set of numbers
numeros = [13, 41, 37, 29, 44]
maximo = 0

for numero in numeros:
    if numero > maximo:
        maximo = numero

print(maximo)
```

## `break`

A `for` loop normally traverses its iterable until completion. Sometimes execution should stop before all elements have been processed. The `break` statement terminates the current loop and transfers execution to the first statement after the loop. An `else` block attached to a `for` or `while` loop is executed only when the loop terminates normally, without encountering `break`.

```python
for i in range(1, 5):
    print(i)
```

```python
for i in range(1, 10):
    print(i)
    if i == 3:
        break
```

```python
for i in range(1, 10):
    if i == 3:
        break
    print(i)
```

```python
n = int(input("Enter an integer: "))
for i in [2, 4, 3, 1, 10, 18, 9, 5, 7]:
    print("Checking element", i)
    if i == n:
        print("Element", i, "found!")
```

## `continue`

The `continue` statement ends the current iteration of a loop and immediately advances execution to the next iteration.

```python
for i in range(1, 6):
    if i == 3:
        continue
    print(i)
```

```python
# Print only the even elements of the list
for i in [2, 4, 7, 1, 0, 8, 9, 5]:
    if i % 2 == 0:
        print(i)
```

```python
# Print only the even elements using continue
for i in [2, 4, 7, 1, 0, 8, 9, 5]:
    if i % 2 == 1:
        continue
    print(i)
```

```python
# Print only the odd elements of the list
for i in [2, 4, 7, 1, 0, 8, 9, 5]:
    if i % 2 == 1:
        print(i)
```

```python
# Print only the odd elements using continue
for i in [2, 4, 7, 1, 0, 8, 9, 5]:
    if i % 2 == 0:
        continue
    print(i)
```

# Nested Loops

```python
# Print a rectangle with n rows and m columns (n × m '*' characters).
n = int(input("Enter the number of rows: "))
m = int(input("Enter the number of columns: "))
for i in range(n):
    print("*" * m)
```

```python
# Print a rectangle with n rows and m columns using nested loops.
n = int(input("Enter the number of rows: "))
m = int(input("Enter the number of columns: "))
for i in range(n):
    for j in range(m):
        print("*", end = "")
    print()
```

```python
# Print an isosceles right triangle with a specified leg length.
n = int(input("Enter the leg length: "))
c = input("Enter the character to use: ")
for i in range(1, n+1):
    print(c * i)
```

```python
# Print an isosceles right triangle using nested loops.
n = int(input("Enter the leg length: "))
c = input("Enter the character to use: ")
for i in range(1, n+1):
    for j in range(i):
        print(c, end = "")
    print()
```

```python
# Repeat the triangle example.
n = int(input("Enter the leg length: "))
c = input("Enter the character to use: ")
for i in range(1, n+1):
    print(c * i)
```

```python
# Print times at one-minute intervals within the selected hour range
for h in range(1):
    for m in range(60):
        print('{:02d}:{:02d}'.format(h, m))
```

```python
# Print times at one-second intervals within the selected hour range
for h in range(1):
    for m in range(60):
        for s in range(60):
            print('{:02d}:{:02d}:{:02d}'.format(h, m, s))
```
