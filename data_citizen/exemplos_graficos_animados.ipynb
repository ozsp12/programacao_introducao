# Raceplot (versão estocástica)

```python
def make_synthetic_race_data(
    countries,
    n_frames=40,
    start_range=(200, 2000),
    drift_range=(50, 250),
    noise_scale=80,
    seed=42,
    date_start="2020-03-01"
):
    rng = np.random.default_rng(seed)
    k = len(countries)

    x0 = rng.uniform(*start_range, size=k)
    drift = rng.uniform(*drift_range, size=k)

    X = np.zeros((n_frames, k))
    X[0] = x0

    for t in range(1, n_frames):
        noise = rng.normal(0, noise_scale, size=k)
        X[t] = np.maximum(0, X[t-1] + drift + noise)

    dates = pd.date_range(date_start, periods=n_frames, freq="D")
    dates = [d.strftime("%Y-%m-%d") for d in dates]

    return X, dates

def animate_bar_race(
    countries,
    colors=None,
    n_frames=40,
    seed=42,
    x_max=None,
    interval=300,
    date_start="2020-03-01"
):
    # dataset sintético
    X, dates = make_synthetic_race_data(
        countries=countries,
        n_frames=n_frames,
        seed=seed,
        date_start=date_start
    )

    k = len(countries)
    if colors is None:
        colors = [None]*k  # matplotlib escolhe padrão

    if x_max is None:
        x_max = float(X.max()) * 1.15

    fig, ax = plt.subplots(figsize=(8, 4))
    ax.set_title("COVID-19 Deaths by Country", fontsize=18)
    ax.xaxis.grid(True, linestyle="--", alpha=0.5)
    ax.set_axisbelow(True)
    ax.set_xlim(0, x_max)

    # barras iniciais
    bars = ax.barh(countries, X[0], color=colors)
    ax.invert_yaxis()

    # labels numéricos
    val_txt = [ax.text(0, 0, "", va="center", fontsize=11) for _ in range(k)]

    # data grande
    date_txt = ax.text(0.65*x_max, k-2, "", fontsize=18)

    def update(frame):
        vals = X[frame]

        # (opcional) ordenar por valor (race de verdade)
        order = np.argsort(vals)
        ordered_c = [countries[i] for i in order]
        ordered_vals = vals[order]
        ordered_colors = [colors[i] for i in order]

        ax.clear()
        ax.set_title("COVID-19 Deaths by Country", fontsize=18)
        ax.xaxis.grid(True, linestyle="--", alpha=0.5)
        ax.set_axisbelow(True)
        ax.set_xlim(0, x_max)

        bars = ax.barh(ordered_c, ordered_vals, color=ordered_colors)
        ax.invert_yaxis()

        for bar, v in zip(bars, ordered_vals):
            ax.text(v + 0.01*x_max, bar.get_y() + bar.get_height()/2,
                    f"{int(v):,}", va="center", fontsize=11)

        ax.text(0.65*x_max, k-2, dates[frame], fontsize=18)
        return []
    plt.close()
    ani = FuncAnimation(fig, update, frames=n_frames, interval=interval, blit=False)
    return HTML(ani.to_jshtml())

# ----------------------------
# EXEMPLO DE USO
# ----------------------------
countries = ["Italy", "Spain", "China", "Iran", "USA", "UK"]
colors = ["limegreen", "#2f3b3b", "dodgerblue", "magenta", "hotpink", "red"]

animate_bar_race(
    countries=countries,
    colors=colors,
    n_frames=40,
    seed=7,
    interval=250,
    date_start="2020-03-01"
)
```

# Série de Fourier animada

```python
# -----------------------------
# Init
# -----------------------------
def init():
    line.set_data([], [])
    scat.set_offsets(np.empty((0, 2)))
    return line, scat

# -----------------------------
# Update: desenha curva + pontos gradualmente
# -----------------------------
def update(i):

    # linha cresce
    line.set_data(x[:i], y[:i])

    # pontos aparecem
    j = int(i * len(xp) / len(x))
    scat.set_offsets(np.c_[xp[:j], yp[:j]])

    return line, scat

# -----------------------------
# Função alvo (como na figura)
# f(x) = x + 10 sin(5x) + 7 cos(4x)
# -----------------------------
def f(x):
    return x + 10*np.sin(5*x) + 7*np.cos(4*x)

# domínio contínuo
x = np.linspace(0, 9, 600)
y = f(x)

# pontos discretos ("indivíduos")
xp = np.linspace(0, 9, 120)
yp = f(xp)

# -----------------------------
# Figura
# -----------------------------
fig, ax = plt.subplots(figsize=(8,5))

# curva (linha) começa vazia
line, = ax.plot([], [], color="orange", lw=2,
                label=r"$f(x)=x+10\sin(5x)+7\cos(4x)$")

# população (scatter) começa vazia
scat = ax.scatter([], [], s=25, color="tomato",
                  label="Individuals in the population")

ax.set_xlim(0, 9)
ax.set_ylim(-16, 26)
ax.set_xlabel("x")
ax.set_ylabel(r"$f(x)$")
ax.legend(loc="upper left")
plt.close()

# -----------------------------
# Animação
# -----------------------------
ani = FuncAnimation(
    fig,
    update,
    frames=len(x),
    init_func=init,
    interval=15,
    blit=True
)

HTML(ani.to_jshtml())
```

```python
# Animation functions
def init():
    pts.set_offsets(np.empty((0, 2)))
    mean_text.set_text("")
    title_text.set_text("")
    ax2.clear()
    ax2.set_xlim(0, 50)
    ax2.set_ylim(0, 0.25)
    ax2.set_title("Distribution of Sample Means", fontsize=12)
    return pts, mean_text, title_text

def update(frame):
    global means

    # --- Sample from exponential distribution ---
    u = np.random.rand(sample_size)
    sample = -20 * np.log(1 - u)

    # Points on the PDF curve
    y_sample = density(sample)

    # Sample mean
    sample_mean = np.mean(sample)
    means.append(sample_mean)

    # Update scatter points
    pts.set_offsets(np.c_[sample, y_sample])

    # Update labels
    title_text.set_text(f"Sample Size = {sample_size}")
    mean_text.set_text(rf"$\bar{{x}}$ = {sample_mean:.2f}")

    # Update histogram panel
    ax2.clear()

    # Only plot histogram once we have enough data
    if len(means) > 1:
        ax2.hist(
            means,
            bins=25,
            range=(0, 50),
            density=True,
            color="purple",
            alpha=0.4
        )

    ax2.set_xlim(0, 50)
    ax2.set_ylim(0, 0.25)
    ax2.set_title("Distribution of Sample Means", fontsize=12)

    return pts, mean_text, title_text
##########################################################################################
# Parameters
np.random.seed(12)

frames = 200
sample_size = 10

# Target density: Exponential-like PDF
def density(x):
    return 0.05 * np.exp(-x / 20)

x_grid = np.linspace(0, 50, 600)
pdf_vals = density(x_grid)

# Storage for sample means
means = []


# Figure layout
fig, (ax1, ax2) = plt.subplots(
    2, 1,
    figsize=(8, 8),
    gridspec_kw={"height_ratios": [1, 1.2]}
)

# --- Top panel: PDF + sampled points ---
ax1.plot(x_grid, pdf_vals, color="purple", lw=2)

pts = ax1.scatter([], [], color="purple", s=35)

mean_text = ax1.text(
    0.52, 0.55, "",
    transform=ax1.transAxes,
    fontsize=14
)

title_text = ax1.text(
    0.55, 0.80, "",
    transform=ax1.transAxes,
    fontsize=16
)

ax1.set_xlim(0, 50)
ax1.set_ylim(0, 0.055)

# --- Bottom panel: histogram of sample means ---
ax2.set_xlim(0, 50)
ax2.set_ylim(0, 0.25)
ax2.set_title("Distribution of Sample Means", fontsize=12)

plt.close()
##########################################################################################
# Build animation
ani = FuncAnimation(
    fig,
    update,
    frames=frames,
    init_func=init,
    interval=120,
    blit=False
)
# Display in Jupyter
HTML(ani.to_jshtml())
```

# Moléculas vibrando

```python
# Update: simple harmonic motion around equilibrium
# x(t) = x0 + A cos(ωt + φ) * u
def update(frame):
    t = 0.05 * frame

    displacement = A * np.cos(omega * t + phi)

    x = x0 + displacement * ux
    y = y0 + displacement * uy

    scat.set_offsets(np.c_[x, y])
    return scat,
##########################################################################################
# Harmonic oscillator vibration (molecules-like motion)
np.random.seed(10)
n_points = 40
# Equilibrium positions (centers of mass)
x0 = np.random.normal(0, 2.5, n_points)
y0 = np.random.normal(0, 2.5, n_points)
sizes = np.random.uniform(120, 260, n_points)
colors = np.random.rand(n_points)
# Harmonic oscillator parameters per particle
A = np.random.uniform(0.05, 0.10, n_points)          # amplitude
omega = np.random.uniform(1.5, 5.5, n_points)        # angular frequency
phi = np.random.uniform(0, 2*np.pi, n_points)        # phase
# Random vibration direction (unit vectors)
theta = np.random.uniform(0, 2*np.pi, n_points)
ux = np.cos(theta)
uy = np.sin(theta)
##########################################################################################
# Figure
fig, ax = plt.subplots(figsize=(6, 4))
ax.set_xlim(-10, 10)
ax.set_ylim(-10, 10)
scat = ax.scatter(
    x0, y0,
    s=sizes,
    c=colors,
    cmap="jet",
    edgecolors="black",
    linewidth=1.2
)
plt.close()
##########################################################################################
# Animation
ani = FuncAnimation(
    fig,
    update,
    frames=400,
    interval=40,
    blit=True
)
#
HTML(ani.to_jshtml())
```

# Spin field animated

```python
# Update: Precession effect (3D illusion)
# Constant magnitude, but projected rotation changes speed
def update(frame):
    t = 0.06 * frame

    # Precession-like modulation
    theta = phase0 + t
    phi   = 0.7*np.sin(0.5*t)  # out-of-plane wobble illusion

    # Projected rotating spins (constant length)
    U = A * np.cos(theta) * np.cos(phi)
    V = A * np.sin(theta) * np.cos(phi)

    quiv.set_UVC(U, V)

    return quiv,
##########################################################################################
# 2D Spin Field with 3D-like Precession Illusion
np.random.seed(20)
nx, ny = 9, 6
x = np.linspace(0.5, 6.5, nx)
y = np.linspace(0.5, 6.5, ny)
X, Y = np.meshgrid(x, y)
# Base spin directions (random initial phases)
phase0 = np.random.uniform(0, 2*np.pi, size=X.shape)
# Arrow base length (slightly larger)
A = 2.25
##########################################################################################
# Figure
fig, ax = plt.subplots(figsize=(7, 4.5))
ax.set_xlim(-1, 7)
ax.set_ylim(-1, 7)
# Initial vectors
U0 = A * np.cos(phase0)
V0 = A * np.sin(phase0)
quiv = ax.quiver(
    X, Y,
    U0, V0,
    color="red",
    angles="xy",
    scale_units="xy",
    scale=3.0,
    width=0.010
)
plt.close()
##########################################################################################
# Animation
ani = FuncAnimation(
    fig,
    update,
    frames=400,
    interval=40,
    blit=True
)
#
HTML(ani.to_jshtml())
```
