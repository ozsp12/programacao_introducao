```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Configuração estética clássica
sns.set(style="whitegrid")
np.random.seed(42)

# Dados sintéticos
x = np.linspace(0, 2*np.pi, 100)
y = np.sin(x)

scatter_x = np.random.rand(80) * 7
scatter_y = 30 - 2*scatter_x + np.random.randn(80)

categories = ["A", "B", "C", "D"]
values = [21, 10, 28, 9]

iris = sns.load_dataset("iris")

heat_data = np.random.rand(10, 10)

pareto_data = np.sort(np.random.randint(5, 50, 6))[::-1]

# Figura 4×3
fig, axes = plt.subplots(4, 3, figsize=(16, 14))
axes = axes.flatten()

# 1. Scatter Plot
axes[0].scatter(scatter_x, scatter_y)
axes[0].set_title("Scatter Plot")

# 2. Bar Plot
axes[1].bar(categories, values)
axes[1].set_title("Bar Plot")

# 3. Histogram with KDE
sns.histplot(scatter_y, kde=True, ax=axes[2])
axes[2].set_title("Histogram with KDE")

# 4. Box Plot
sns.boxplot(x="species", y="sepal_length", data=iris, ax=axes[3])
axes[3].set_title("Box Plot")

# 5. Line Plot
axes[4].plot(x, y)
axes[4].set_title("Line Plot")

# 6. Heatmap
sns.heatmap(heat_data, ax=axes[5], cmap="rocket")
axes[5].set_title("Heatmap")

# 7. Violin Plot
sns.violinplot(x="species", y="sepal_length", data=iris, ax=axes[6])
axes[6].set_title("Violin Plot")

# 8. Area Plot
axes[7].fill_between(x, y, alpha=0.6)
axes[7].set_title("Area Plot")

# 9. Bubble Plot
sizes = np.random.rand(80) * 200
axes[8].scatter(scatter_x, scatter_y, s=sizes, alpha=0.5)
axes[8].set_title("Bubble Plot")

# 10. Density Plot
sns.kdeplot(iris["sepal_length"], fill=True, ax=axes[9])
axes[9].set_title("Density Plot")

# 11. Faceted Plot (simplificado)
sns.scatterplot(
    x="sepal_width",
    y="petal_width",
    hue="species",
    data=iris,
    ax=axes[10]
)
axes[10].set_title("Faceted Plot")

# 12. Pareto Plot
axes[11].bar(range(len(pareto_data)), pareto_data)
axes[11].set_title("Pareto Plot")

# Ajuste final
plt.tight_layout()
plt.show()
```
