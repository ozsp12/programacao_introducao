# Lambdas

A lambda expression is most useful as a compact alternative to `def` when a small executable expression must be inserted where a function declaration would be syntactically inconvenient.

## Comparing lambda expressions and functions

```python
def func(x, y, z):
    return x + y + z

func(2, 3, 4)
```

```python
f = lambda x, y, z: x + y + z

f(2, 3, 4)
```

The code below builds a list of three functions by embedding lambda expressions inside a list literal. A `def` statement cannot appear directly inside a list literal because it is a statement rather than an expression. The equivalent implementation with `def` therefore requires temporary function names and definitions outside the immediate context in which the functions are used.

```python
L = [lambda x: x ** 2, lambda x: x ** 3, lambda x: x ** 4]

for f in L:
    print(f(2))
```

```python
# Returns the value of the first lambda using the number inside parentheses as its argument
L[0](2)
```

```python
# Returns the value of the second lambda using the number inside parentheses as its argument
L[1](2)
```

```python
# Returns the value of the third lambda using the number inside parentheses as its argument
L[2](2)
```

```python
def f1(x):
    return x ** 2

def f2(x):
    return x ** 3

def f3(x):
    return x ** 4

R = [f1, f2, f3]

for f in R:
    print(f(3))
```

```python
# Returns the value of the first function using the number inside parentheses as its argument
R[0](3)
```

```python
# Returns the value of the second function using the number inside parentheses as its argument
R[1](3)
```

```python
# Returns the value of the third function using the number inside parentheses as its argument
R[2](3)
```
